```python



SOURCE_DIRECTORY = "/Users/pgpetitmangin/underwater/Underwater_Software/SLAM/data_sortie_mer/frames/gauche" 

MAIN_OUTPUT_DIRECTORY = "/Users/pgpetitmangin/underwater/Underwater_Software/SLAM/data_sortie_mer/frames/gauche"

# Resizing 
TARGET_WIDTH = 320  
TARGET_HEIGHT = 240 

# Extraction frequency
FRAME_STEP = 1
VIDEO_EXTENSION = "*.mp4" 

# Video Extraction Configuration
VIDEO_CONFIG = {
    'width': 1920,      
    'height': 1080,     
    'step': 1          
}





```

```python
def extract_frames_from_single_video(video_path : str,
                                    frames_output_dir : str ,
                                    target_width=None,
                                    target_height=None,
                                    frame_step=1) -> int :
    """
    Uses Open CV to extract the frames from a single video.
    """
    # Create the frames output directory
    if not os.path.exists(frames_output_dir):
        os.makedirs(frames_output_dir)
        
    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        return 0

    frame_count = 0
    saved_frames_count = 0
    
    while True:
        ret, frame = cap.read()
        if not ret:
            break

        if frame_count % frame_step == 0:
            frame_to_process = frame
            
            # Resizing
            if target_width is not None and target_height is not None:
                dim = (target_width, target_height)
                frame_to_process = cv2.resize(frame_to_process, dim, interpolation=cv2.INTER_AREA)

            # Save frame
            file_name = os.path.join(frames_output_dir, f"frame_{saved_frames_count:06d}.jpg")
            cv2.imwrite(file_name, frame_to_process)
            saved_frames_count += 1

        frame_count += 1
        print(frame_to_process.shape)     
    cap.release()
    return saved_frames_count
```

```python
def batch_process_video_and_audio(source_dir : str,
                                  main_output_dir : str, 
                                  video_config : str,
                                  video_extension="*.mp4") -> None :
    """
    Processes all video files in a directory: extracts both frames and audio.
    """
    print(f"--- Starting Batch Processing (Audio + Video) from: {source_dir} ---")
    
    if not os.path.exists(main_output_dir):
        os.makedirs(main_output_dir)

    # Find all videos matching the extension
    video_files = glob.glob(os.path.join(source_dir, video_extension))
    
    if not video_files:
        print(f"No video files found in {source_dir} with extension {video_extension}")
        return

    total_videos = len(video_files)
    processed_count = 0

    for video_path in video_files:
        video_filename = os.path.basename(video_path)
        
        # Create a unique sub-folder for this video (e.g., main_output_dir/video_name)
        video_name_no_ext = os.path.splitext(video_filename)[0]
        video_output_dir = os.path.join(main_output_dir, video_name_no_ext)
        
        # Specific sub-folders for frames and audio
        frames_output_dir = os.path.join(video_output_dir, "frames")
        
        
        print(f"\n[{processed_count + 1}/{total_videos}] Processing: {video_filename}")
        
        
        # --- A. Frame Extraction (Video) ---
        frames_saved = extract_frames_from_single_video(
            video_path, 
            frames_output_dir, 
            video_config['width'], 
            video_config['height'], 
            video_config['step']
        )
        print(f"   -> Frames saved: {frames_saved}")

        

        processed_count += 1

    print("\n--- Batch Processing Complete ---")
    print(f"Total videos processed: {processed_count}")
```
