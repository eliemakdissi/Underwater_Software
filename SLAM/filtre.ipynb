```python
import cv2
import os
```

```python



left_input = "/Users/pgpetitmangin/underwater/Underwater_Software/SLAM/data_sortie_mer/frames/gauche/sortie_left.mp4_fixed/frames"
right_input = "/Users/pgpetitmangin/underwater/Underwater_Software/SLAM/data_sortie_mer/frames/droite/sortie_right.mp4_fixed/frames"

left_output = "/Users/pgpetitmangin/underwater/Underwater_Software/SLAM/test_frames_mer_1/gauche"
right_output = "/Users/pgpetitmangin/underwater/Underwater_Software/SLAM/test_frames_mer_1/droite"

os.makedirs(left_output, exist_ok=True)
os.makedirs(right_output, exist_ok=True)

def blur_score(image):
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    return cv2.Laplacian(gray, cv2.CV_64F).var()

threshold = 120

left_files = sorted([f for f in os.listdir(left_input) if f.endswith(".jpg")])

kept = 0
skipped = 0

for filename in left_files:
    path_left = os.path.join(left_input, filename)
    path_right = os.path.join(right_input, filename)

    # si la frame droite n'existe pas → on skip
    if not os.path.exists(path_right):
        skipped += 1
        continue

    img_left = cv2.imread(path_left)
    img_right = cv2.imread(path_right)

    if img_left is None or img_right is None:
        continue

    score_left = blur_score(img_left)
    score_right = blur_score(img_right)

    if score_left >= threshold and score_right >= threshold:
        cv2.imwrite(os.path.join(left_output, filename), img_left)
        cv2.imwrite(os.path.join(right_output, filename), img_right)
        kept += 1

print(f"Frames gardées : {kept}")
print(f"Frames ignorées (pas de paire) : {skipped}")
```
